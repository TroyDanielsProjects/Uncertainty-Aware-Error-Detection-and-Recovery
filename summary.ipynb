```python
import sqlite3
import warnings
import re
import textwrap
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
from sklearn.metrics import pairwise_distances_argmin_min
from IPython.display import display, Markdown

warnings.simplefilter('ignore')
pd.set_option('display.max_rows', 500)
pd.set_option('display.float_format', '{:.4f}'.format)
sns.set_theme(style="white", context="paper", font_scale=1.2)

DB_PATH = 'db/results.sqlite'

CLUSTER_CONFIG = {
    0: ("High Mech Unc", "These samples have slightly elevated mechanistic uncertainty, but are by and large ok."),
    1: ("High Logit Gap", "High confidence on standard, straightforward problems."),
    2: ("High Sem Unc", "Elevated semantic uncertainty, but ok performance."),
    3: ("Max Heuristic", "A lot of very confident language."),
    4: ("Low Mech Unc", "Rote calculations with low mechanistic uncertainty."),
    5: ("High Mech + High Logit Gap", "High mechanical uncertainty, but with high logit gap / confident results"),
    6: ("Min Mech Unc", "Easier problems solved with minimal mechanistic uncertainty."),
    7: ("Max Logit Gap", "Extreme logit gap confidence likely due to compounding trivial steps."),
    8: ("Baseline", "Average performance across all uncertainty metrics."),
    9: ("Max Sem", "Catastrophically high semantic uncertainty."),
    10: ("High Sem Unc", "Very high semantic uncertainty."),
    11: ("Very High Logit Gap", "Extremely confident logit gap results."),
    12: ("High Heuristic", "A lot of confident language.")
}

METRIC_MAP = {
    'uq_mech_score_RL': 'Mech Unc', 
    'uq_semantic_entropy_RL': 'Sem Unc',
    'uq_min_logit_gap_RL': 'Logit Gap', 
    'uq_heuristic_score_RL': 'Heuristic'
}

def sanitize(t): return re.sub(r'\\boxed\{(.*?)\}', r'\1', str(t)).replace('\\', '').replace('$', '\\$')

def load_data():
    with sqlite3.connect(DB_PATH) as conn:
        exp = pd.read_sql("SELECT * FROM Experiments ORDER BY run_date DESC LIMIT 1", conn)
        eid = exp.iloc[0].get('id', exp.iloc[0].get('experiment_id'))
        res = pd.read_sql(f"SELECT * FROM Results WHERE experiment_id = {eid}", conn)
        models = pd.read_sql("SELECT * FROM Models", conn)
        
        df = pd.merge(res, models, left_on='model_id' if 'model_id' in res else 'model_id', right_on='id' if 'id' in models else 'model_id', how='left')
        df['training_method'] = df.get('training_method', df.get('type', 'Unknown'))
        df['qid'] = df.get('question_id_external', df.get('question_id', df.index))
        
        merged = df[df['training_method'] == 'SFT'].set_index('qid').join(df[df['training_method'] == 'RL'].set_index('qid'), lsuffix='_SFT', rsuffix='_RL', how='inner')
        t_col = next((c for c in ['full_trace_text_RL', 'full_trace_text', 'generated_text_RL'] if c in merged.columns), 'generated_text_RL')
        
        merged['trace_rl'] = merged[t_col].astype(str)
        merged['pred_rl'] = merged['predicted_answer_RL'].astype(str)
        merged['gold_ans'] = merged['gold_answer_RL'].astype(str)
        return merged

def run_report():
    df = load_data()
    if df is None or df.empty: return

    metrics = [k for k in METRIC_MAP if k in df.columns]
    g_stats = {m: {'med': df[m].median(), 'std': df[m].std()} for m in metrics}
    
    X = StandardScaler().fit_transform(df[metrics].fillna(df[metrics].mean()))
    kmeans = KMeans(n_clusters=13, random_state=42, n_init=10)
    df['cluster'] = kmeans.fit_predict(X)
    
    tsne = TSNE(n_components=2, perplexity=min(20, len(df)-1), random_state=42, init='pca').fit_transform(X)
    df['tsne_x'], df['tsne_y'] = tsne[:, 0], tsne[:, 1]

    display(Markdown("# Clustering Summary"))
    display(Markdown(f"**Total Samples:** {len(df)}"))
    display(Markdown("---"))

    for cid in sorted(df['cluster'].unique(), key=lambda x: df[df['cluster']==x]['is_correct_RL'].mean(), reverse=True):
        sub = df[df['cluster'] == cid]
        name, desc = CLUSTER_CONFIG.get(cid, ("Unknown", "No description available."))
        acc = sub['is_correct_RL'].mean()
        size_pct = len(sub) / len(df)
        
        table_rows = []
        for m in metrics:
            local_med = sub[m].median()
            glb_med = g_stats[m]['med']
            glb_std = g_stats[m]['std']
            z_score = (local_med - glb_med) / glb_std if glb_std > 0 else 0
            
            table_rows.append({
                "Metric": METRIC_MAP[m],
                "Cluster Med": f"{local_med:.3f}",
                "Global Med": f"{glb_med:.3f}",
                "Deviation": f"{z_score:+.1f}σ",
                "90% Interval": f"[{sub[m].quantile(0.05):.2f} - {sub[m].quantile(0.95):.2f}]"
            })
        
        display(Markdown(f"## Cluster {cid}: {name}"))
        display(Markdown(f"**Accuracy:** {acc:.1%} | **Size:** {len(sub)} samples ({size_pct:.1%})"))
        display(Markdown(f"> *{desc}*"))
        display(pd.DataFrame(table_rows))

        center = kmeans.cluster_centers_[cid]
        closest = pairwise_distances_argmin_min([center], X[df['cluster'] == cid])[0][0]
        arch_idx = sub.index[closest]
        examples = [arch_idx] + list(sub.sample(min(2, len(sub)-1)).index) if len(sub) > 1 else [arch_idx]
        
        display(Markdown("### Representative Examples"))
        for i, idx in enumerate(list(dict.fromkeys(examples))):
            row = df.loc[idx]
            trace_preview = textwrap.shorten(row['trace_rl'], width=1000, placeholder="...")
            
            md_ex = f"""
**Example {i+1}**
* **Gold:** `{row['gold_ans']}` | **Pred:** `{row['pred_rl']}`
* **Trace:** {sanitize(trace_preview)}
            """
            display(Markdown(md_ex))
            
            ex_stats = []
            for m in metrics:
                val = row[m]
                z = (val - g_stats[m]['med']) / g_stats[m]['std'] if g_stats[m]['std'] > 0 else 0
                ex_stats.append(f"{METRIC_MAP[m]}: {val:.3f} ({z:+.1f}σ)")
            
            display(Markdown(f"`{' | '.join(ex_stats)}`"))
        
        display(Markdown("---"))

    plt.figure(figsize=(20, 15))
    cluster_labels = df['cluster'].map(lambda c: f"{CLUSTER_CONFIG.get(c, ('Unknown',))[0]} ({df[df['cluster']==c]['is_correct_RL'].mean():.0%})")
    sns.scatterplot(data=df.sort_values('cluster'), x='tsne_x', y='tsne_y', hue=cluster_labels, style=cluster_labels, palette='turbo', s=150, alpha=0.9, edgecolor='k')
    
    for cid in df['cluster'].unique():
        sub = df[df['cluster'] == cid]
        name = CLUSTER_CONFIG.get(cid, ("Unknown",))[0]
        acc = sub['is_correct_RL'].mean()
        
        center = kmeans.cluster_centers_[cid]
        closest = pairwise_distances_argmin_min([center], X[df['cluster'] == cid])[0][0]
        arch_ans = df.loc[sub.index[closest], 'pred_rl']
        
        color = '#ffe0e0' if acc < 0.6 else ('#fffacd' if "Heuristic" in name else '#e0fffa')
        plt.text(sub['tsne_x'].mean(), sub['tsne_y'].mean(), f"C{cid}: {name}\nAcc: {acc:.0%}\nAns: {arch_ans}", 
                 fontsize=8, fontweight='bold', ha='center', bbox=dict(facecolor=color, alpha=0.8, boxstyle='round,pad=0.4'))

    plt.title("Clusters")
    plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    run_report()
```

```python
import sqlite3
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')
sns.set_context("paper", font_scale=1.2)
sns.set_style("whitegrid")

# Robust DB Path Detection
# Check current directory and a 'db' subdirectory
possible_paths = ['db/results.sqlite']
DB_PATH = next((p for p in possible_paths if os.path.exists(p)), None)

METRIC_MAP = {
    'uq_mech_score': 'Mech Unc',
    'uq_semantic_entropy': 'Sem Unc',
    'uq_min_logit_gap': 'Logit Gap',
    'uq_heuristic_score': 'Heuristic'
}

def load_data(model_label):
    if not DB_PATH: raise FileNotFoundError("Database not found.")
    
    with sqlite3.connect(DB_PATH) as conn:
        try:
            exp_id = conn.execute("SELECT experiment_id FROM Experiments ORDER BY run_date DESC LIMIT 1").fetchone()[0]
            model_res = conn.execute(f"SELECT model_id FROM Models WHERE training_method = '{model_label}'").fetchone()
            if not model_res: return None
            
            df = pd.read_sql_query(f"""
                SELECT is_correct, uq_mech_score, uq_semantic_entropy, uq_min_logit_gap, uq_heuristic_score
                FROM Results WHERE experiment_id = {exp_id} AND model_id = {model_res[0]}
            """, conn)
            
            df.rename(columns=METRIC_MAP, inplace=True)
            df["Outcome"] = df["is_correct"].map({1: "Correct", 0: "Incorrect"})
            return df
        except: return None

def plot_distributions():
    df_sft = load_data('SFT')
    df_rl = load_data('RL')
    
    if df_sft is None or df_rl is None:
        print("Could not load data for both models.")
        return

    metrics = ["Mech Unc", "Sem Unc", "Logit Gap", "Heuristic"]
    fig, axes = plt.subplots(2, 4, figsize=(24, 10))
    
    # Plot SFT Row
    for i, m in enumerate(metrics):
        if m in df_sft.columns:
            sns.kdeplot(data=df_sft, x=m, hue="Outcome", fill=True, common_norm=False, 
                        palette={"Correct": "#2ecc71", "Incorrect": "#e74c3c"}, ax=axes[0,i], alpha=0.3)
            axes[0,i].set_title(f"SFT: {m}")
            axes[0,i].set_xlabel("")
            
    # Plot RL Row
    for i, m in enumerate(metrics):
        if m in df_rl.columns:
            sns.kdeplot(data=df_rl, x=m, hue="Outcome", fill=True, common_norm=False, 
                        palette={"Correct": "#2ecc71", "Incorrect": "#e74c3c"}, ax=axes[1,i], alpha=0.3)
            axes[1,i].set_title(f"RL: {m}")
            
    plt.suptitle("Uncertainty Distributions by Correctness: SFT vs RL", y=1.02, fontsize=16)
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    if DB_PATH:
        plot_distributions()
    else:
        print("Database not found.")
```
