```python
import sqlite3
import warnings
import re
import textwrap
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
from sklearn.metrics import pairwise_distances_argmin_min
from IPython.display import display, Markdown

warnings.simplefilter('ignore')
pd.set_option('display.max_rows', 500)
pd.set_option('display.float_format', '{:.4f}'.format)
sns.set_theme(style="white", context="paper", font_scale=1.2)

DB_PATH = 'db/results.sqlite'

BASE_CLUSTER_NAMES = {
    0: "High Mech Unc", 1: "High Logit Gap",
    2: "High Sem Unc", 3: "Max Heuristic ",
    4: "Low Mech Unc", 5: "High Mech + High Logit Gap",
    6: "Min Mech Unc", 7: "Max Logit Gap",
    8: "Baseline", 9: "Max Sem",
    10: "High Sem Unc", 11: "Very High Logit Gap",
    12: "High Heuristic"
}

METRIC_MAP = {
    'uq_mech_score_RL': 'Mech Unc', 'uq_semantic_entropy_RL': 'Sem Unc',
    'uq_min_logit_gap_RL': 'Logit Gap', 'uq_heuristic_score_RL': 'Heuristic'
}

def sanitize(t): return re.sub(r'\\boxed\{(.*?)\}', r'\1', str(t)).replace('\\', '').replace('$', '\\$')
def get_z(v, m, s): return "+0.0σ" if s == 0 else f"{(v-m)/s:+.1f}σ"

def load_data():
    with sqlite3.connect(DB_PATH) as conn:
        exp = pd.read_sql("SELECT * FROM Experiments ORDER BY run_date DESC LIMIT 1", conn)
        eid = exp.iloc[0].get('id', exp.iloc[0].get('experiment_id'))
        res = pd.read_sql(f"SELECT * FROM Results WHERE experiment_id = {eid}", conn)
        models = pd.read_sql("SELECT * FROM Models", conn)
        
        df = pd.merge(res, models, left_on='model_id' if 'model_id' in res else 'model_id', right_on='id' if 'id' in models else 'model_id', how='left')
        df['training_method'] = df.get('training_method', df.get('type', 'Unknown'))
        df['qid'] = df.get('question_id_external', df.get('question_id', df.index))
        
        merged = df[df['training_method'] == 'SFT'].set_index('qid').join(df[df['training_method'] == 'RL'].set_index('qid'), lsuffix='_SFT', rsuffix='_RL', how='inner')
        t_col = next((c for c in ['full_trace_text_RL', 'full_trace_text', 'generated_text_RL'] if c in merged.columns), 'generated_text_RL')
        
        merged['trace_rl'] = merged[t_col].astype(str)
        merged['pred_rl'] = merged['predicted_answer_RL'].astype(str)
        merged['gold_ans'] = merged['gold_answer_RL'].astype(str)
        return merged

def run_report():
    df = load_data()
    if df is None or df.empty: return

    metrics = [k for k in METRIC_MAP if k in df.columns]
    g_stats = {m: {'med': df[m].median(), 'std': df[m].std()} for m in metrics}
    
    X = StandardScaler().fit_transform(df[metrics].fillna(df[metrics].mean()))
    kmeans = KMeans(n_clusters=13, random_state=42, n_init=10)
    df['cluster'] = kmeans.fit_predict(X)
    
    tsne = TSNE(n_components=2, perplexity=min(20, len(df)-1), random_state=42, init='pca').fit_transform(X)
    df['tsne_x'], df['tsne_y'] = tsne[:, 0], tsne[:, 1]

    display(Markdown("# Cognitive State Analysis Report"))
    display(Markdown("---"))

    for cid in sorted(df['cluster'].unique(), key=lambda x: df[df['cluster']==x]['is_correct_RL'].mean(), reverse=True):
        sub = df[df['cluster'] == cid]
        name = BASE_CLUSTER_NAMES.get(cid, "Unknown")
        acc = sub['is_correct_RL'].mean()
        size_pct = len(sub) / len(df)
        
        # Profile Summary
        z_scores = [f"**{METRIC_MAP[m]}**: {get_z(sub[m].median(), g_stats[m]['med'], g_stats[m]['std'])}" for m in metrics]
        
        # Statistics Table
        stats_data = []
        for m in metrics:
            stats_data.append({
                "Metric": METRIC_MAP[m],
                "Cluster Med": f"{sub[m].median():.3f}",
                "Global Med": f"{g_stats[m]['med']:.3f}",
                "Mean": f"{sub[m].mean():.3f}",
                "Std Dev": f"{sub[m].std():.3f}",
                "90% Range": f"[{sub[m].quantile(0.05):.3f} - {sub[m].quantile(0.95):.3f}]"
            })
        stats_df = pd.DataFrame(stats_data)

        # Header
        display(Markdown(f"## Cluster {cid}: {name}"))
        display(Markdown(f"**Accuracy:** {acc:.1%} | **Size:** N={len(sub)} ({size_pct:.1%})"))
        display(Markdown(" | ".join(z_scores)))
        display(stats_df)

        # Examples
        center = kmeans.cluster_centers_[cid]
        closest = pairwise_distances_argmin_min([center], X[df['cluster'] == cid])[0][0]
        arch_idx = sub.index[closest]
        examples = [arch_idx] + list(sub.sample(min(2, len(sub)-1)).index) if len(sub) > 1 else [arch_idx]
        
        display(Markdown("### Representative Examples"))
        for i, idx in enumerate(list(dict.fromkeys(examples))):
            row = df.loc[idx]
            trace_preview = textwrap.shorten(row['trace_rl'], width=1000, placeholder="...")
            
            md_ex = f"""
**Example {i+1}**
* **Gold:** `{row['gold_ans']}`
* **Pred:** `{row['pred_rl']}`
* **Trace:**
> {sanitize(trace_preview)}
            """
            display(Markdown(md_ex))
            
            # Per-example metrics
            ex_stats = [f"{METRIC_MAP[m]}: {row[m]:.3f} ({get_z(row[m], g_stats[m]['med'], g_stats[m]['std'])})" for m in metrics]
            display(Markdown(f"`{' | '.join(ex_stats)}`"))
        
        display(Markdown("---"))

    # Visualization
    plt.figure(figsize=(20, 15))
    cluster_labels = df['cluster'].map(lambda c: f"{BASE_CLUSTER_NAMES.get(c, 'Unknown')} ({df[df['cluster']==c]['is_correct_RL'].mean():.0%})")
    sns.scatterplot(data=df.sort_values('cluster'), x='tsne_x', y='tsne_y', hue=cluster_labels, style=cluster_labels, palette='turbo', s=150, alpha=0.9, edgecolor='k')
    
    for cid in df['cluster'].unique():
        sub = df[df['cluster'] == cid]
        name = BASE_CLUSTER_NAMES.get(cid, "Unknown")
        acc = sub['is_correct_RL'].mean()
        center = kmeans.cluster_centers_[cid]
        closest = pairwise_distances_argmin_min([center], X[df['cluster'] == cid])[0][0]
        arch_ans = df.loc[sub.index[closest], 'pred_rl']
        
        color = '#ffe0e0' if acc < 0.6 else ('#fffacd' if "Heuristic" in name else '#e0fffa')
        plt.text(sub['tsne_x'].mean(), sub['tsne_y'].mean(), f"C{cid}: {name}\nAcc: {acc:.0%}\nAns: {arch_ans}", 
                 fontsize=8, fontweight='bold', ha='center', bbox=dict(facecolor=color, alpha=0.8, boxstyle='round,pad=0.4'))

    plt.title("Cognitive State Map: Performance & Uncertainty Deviation")
    plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    run_report()
```

```python
import sqlite3
import pandas as pd
import numpy as np
import warnings
import textwrap
import re
from io import StringIO
from difflib import SequenceMatcher

# Advanced Analytics
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
from sklearn.metrics import pairwise_distances_argmin_min

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration
warnings.simplefilter(action='ignore')
pd.set_option('display.max_rows', 500)
pd.set_option('display.width', 1000)
pd.options.display.float_format = '{:.4f}'.format

DB_PATH = 'db/results.sqlite'
sns.set_theme(style="white", context="paper", font_scale=1.2)

# ==============================================================================
#  CONFIG: TAXONOMY & METRICS
# ==============================================================================
BASE_CLUSTER_NAMES = {
    # Mech +0.7σ | Conf -0.2σ
    0:  "High Mech Unc",    

    # Conf +1.2σ | Mech -0.3σ
    1:  "High Logit Gap",         

    # Sem +2.1σ | Mech -0.4σ
    2:  "High Sem Unc",       

    # Heur +6.0σ | Conf +0.5σ
    3:  "Max Heuristic",      

    # Mech -0.7σ | Conf -0.2σ
    4:  "Low Mech Unc",           

    # Mech +1.2σ | Conf +1.1σ
    5:  "High Mech Unc + High Logit Gap",         

    # Mech -2.3σ (Lowest in dataset)
    6:  "Min Mech Unc",           

    # Conf +7.3σ (Highest in dataset)
    7:  "Max Logit Gap",          

    # All metrics ~0.0σ
    8:  "Baseline",              

    # Sem +5.5σ (Highest in dataset) | Mech -3.2σ
    9:  "Max Sem Unc",          

    # Sem +3.9σ | Mech -0.9σ
    10: "High Sem Unc",           

    # Conf +3.8σ | Mech +1.0σ
    11: "Very High Logit Gap",            

    # Heur +4.0σ | Mech +0.0σ
    12: "High Heuristic"              
}

METRIC_MAP = {
    'uq_mech_score_RL': 'Mech Unc',
    'uq_semantic_entropy_RL': 'Sem Unc',
    'uq_min_logit_gap_RL': 'Logit Gap',
    'uq_heuristic_score_RL': 'Heuristic'
}

# ==============================================================================
#  HELPER FUNCTIONS
# ==============================================================================

def sanitize_for_plot(text):
    """Cleans text to prevent Matplotlib MathText errors."""
    if not isinstance(text, str): text = str(text)
    text = re.sub(r'\\boxed\{(.*?)\}', r'\1', text)
    text = text.replace('\\', '')
    text = text.replace('$', '\\$') 
    return text

def get_relative_z_string(val, global_med, global_std):
    if global_std == 0: return "+0.0σ"
    z = (val - global_med) / global_std
    return f"{z:+.1f}σ"

def get_stats_string(series, global_series):
    med_val = series.median()
    global_med = global_series.median()
    mean_val = series.mean()
    std_val = series.std()
    p05 = series.quantile(0.05)
    p95 = series.quantile(0.95)
    
    return (f"Med: {med_val:6.3f} (Glb: {global_med:5.3f}) | "
            f"Mean: {mean_val:6.3f} | "
            f"Std: {std_val:5.3f} | "
            f"90%: [{p05:5.3f} - {p95:5.3f}]")

def wrap_text_full(text, width=100, indent="    "):
    """Wraps text without truncation for report"""
    if not isinstance(text, str): text = str(text)
    wrapper = textwrap.TextWrapper(width=width, initial_indent=indent, subsequent_indent=indent)
    return wrapper.fill(text)

def wrap_plot_text(text, width=50):
    """Wraps text for plot annotations (NO TRUNCATION)"""
    if not isinstance(text, str): text = str(text)
    clean_text = sanitize_for_plot(text)
    wrapper = textwrap.TextWrapper(width=width)
    return "\n".join(wrapper.wrap(clean_text))

# ==============================================================================
#  DATA LOADING
# ==============================================================================
def load_data():
    conn = sqlite3.connect(DB_PATH)
    try:
        exp = pd.read_sql("SELECT * FROM Experiments ORDER BY run_date DESC LIMIT 1", conn)
        exp_id = exp.iloc[0]['id'] if 'id' in exp.columns else exp.iloc[0]['experiment_id']
        res = pd.read_sql(f"SELECT * FROM Results WHERE experiment_id = {exp_id}", conn)
        models = pd.read_sql("SELECT * FROM Models", conn)
        
        m_col = 'model_id' if 'model_id' in res.columns else 'model_id' 
        r_col = 'id' if 'id' in models.columns else ('model_id' if 'model_id' in models.columns else models.columns[0])
            
        df = pd.merge(res, models, left_on=m_col, right_on=r_col, how='left')
        if 'training_method' not in df.columns: df['training_method'] = df.get('type', 'Unknown')
        df['qid'] = df.get('question_id_external', df.get('question_id', df.index))
        
        sft = df[df['training_method'] == 'SFT'].set_index('qid')
        rl = df[df['training_method'] == 'RL'].set_index('qid')
        merged = sft.join(rl, lsuffix='_SFT', rsuffix='_RL', how='inner')
        
        # Priority for Full Trace Text (Reasoning + Answer)
        if 'full_trace_text_RL' in merged.columns:
            target_col = 'full_trace_text_RL'
        elif 'full_trace_text' in merged.columns: 
            target_col = 'full_trace_text'
        elif 'predicted_answer_RL' in merged.columns:
            target_col = 'predicted_answer_RL'
        else:
            target_col = 'generated_text_RL'
            
        merged['ans_rl'] = merged[target_col].astype(str)
        
        return merged
    finally:
        conn.close()

# ==============================================================================
#  CORE ANALYSIS
# ==============================================================================
def run_final_cluster_report():
    df = load_data()
    if df is None or len(df) == 0: return

    # 1. Select Metrics
    metrics = [k for k in METRIC_MAP.keys() if k in df.columns]
    
    # Global Stats
    global_stats = {}
    for m in metrics:
        global_stats[m] = {
            'median': df[m].median(),
            'std': df[m].std()
        }

    # 2. Clustering (Deterministic K=13)
    X = df[metrics].fillna(df[metrics].mean())
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    k = 13
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    df['cluster'] = kmeans.fit_predict(X_scaled)
    
    # 3. Metadata Generation
    cluster_meta = {}
    for cid in sorted(df['cluster'].unique()):
        subset = df[df['cluster'] == cid]
        acc = subset['is_correct_RL'].mean()
        base_name = BASE_CLUSTER_NAMES.get(cid, "Unknown")
        
        
        # Z-Scores for label
        z_scores = {}
        for m_key, m_name in METRIC_MAP.items():
            z_scores[m_name] = get_relative_z_string(
                subset[m_key].median(), 
                global_stats[m_key]['median'], 
                global_stats[m_key]['std']
            )

        # Get Archetype ANSWER for Plotting
        center = kmeans.cluster_centers_[cid]
        subset_X = X_scaled[df['cluster'] == cid]
        closest, _ = pairwise_distances_argmin_min([center], subset_X)
        archetype_row = df.loc[subset.index[closest[0]]]
        
        # Full wrapped answer for plot (NO TRUNCATION)
        plot_a = wrap_plot_text(archetype_row['ans_rl'], width=50)

        cluster_meta[cid] = {
            'name': base_name,
            'acc': acc,
            'z_scores': z_scores,
            'label': f"{base_name} ({acc:.0%})",
            'example_a': plot_a
        }

    df['cluster_label'] = df['cluster'].apply(lambda x: cluster_meta[x]['label'])
    
    # 4. t-SNE
    tsne = TSNE(n_components=2, perplexity=min(20, len(df)-1), random_state=42, init='pca', learning_rate='auto')
    tsne_res = tsne.fit_transform(X_scaled)
    df['tsne_x'] = tsne_res[:, 0]
    df['tsne_y'] = tsne_res[:, 1]

    # ==========================================================================
    #  DETAILED STATISTICAL REPORT
    # ==========================================================================
    print("\n" + "="*80)
    print("="*80)

    sorted_cids = sorted(cluster_meta.keys(), key=lambda x: cluster_meta[x]['acc'], reverse=True)

    for cid in sorted_cids:
        meta = cluster_meta[cid]
        subset = df[df['cluster'] == cid]
        z = meta['z_scores']
        
        print(f"\ncluster {cid}: {meta['name']} ({meta['acc']:.1%})")
        print(f"Stats: Mech {z['Mech Unc']} | Sem {z['Sem Unc']} | Conf {z['Logit Gap']} | Heur {z['Heuristic']}")
        print(f"{'-'*80}")
        print(f"Size: {len(subset)} ({len(subset)/len(df):.1%})")
        
        print("\nstatistics:")
        for m in metrics:
            readable = METRIC_MAP[m]
            stats_str = get_stats_string(subset[m], df[m])
            print(f"  • {readable:<15}: {stats_str}")
        
        print("\n" + "-"*80)
        print(" Representative Examples")
        print("-" * 80)
        
        # 1. Archetype
        center = kmeans.cluster_centers_[cid]
        subset_X = X_scaled[df['cluster'] == cid]
        closest, _ = pairwise_distances_argmin_min([center], subset_X)
        archetype_idx = subset.index[closest[0]]
        
        # 2. Random Samples
        potential_others = subset.index.tolist()
        if archetype_idx in potential_others: potential_others.remove(archetype_idx)
        
        examples_to_show = [archetype_idx]
        if len(potential_others) > 0:
            others = np.random.choice(potential_others, size=min(2, len(potential_others)), replace=False)
            examples_to_show.extend(others)
            
        for i, idx in enumerate(examples_to_show):
            row = df.loc[idx]
            role = f"example {i}"
            
            print(f"\n[{role}]")
            # ONLY PRINTING ANSWER AS REQUESTED
            print(f"A: {wrap_text_full(row['ans_rl'])}")
            
            # Instance Metrics
            z_strs = []
            for m in metrics:
                val = row[m]
                z_val = (val - global_stats[m]['median']) / (global_stats[m]['std'] + 1e-9)
                m_name = METRIC_MAP[m]
                z_strs.append(f"{m_name}: {val:.3f} ({z_val:+.1f}σ)")
            
            print(f"   >>> {' | '.join(z_strs)}")

        print(f"{'='*80}")

    # ==========================================================================
    #  VISUALIZATION
    # ==========================================================================
    plt.figure(figsize=(24, 18))
    df.sort_values(['cluster'], inplace=True)
    
    sns.scatterplot(
        data=df, 
        x='tsne_x', 
        y='tsne_y', 
        hue='cluster_label',
        style='cluster_label', 
        palette='turbo', 
        s=150, 
        alpha=0.9,
        edgecolor='k'
    )
    
    for cid in df['cluster'].unique():
        meta = cluster_meta[cid]
        c_subset = df[df['cluster'] == cid]
        cx, cy = c_subset['tsne_x'].mean(), c_subset['tsne_y'].mean()
        z = meta['z_scores']
        
        # Color coding
        bg_color = 'white'
        if meta['acc'] < 0.6: bg_color = '#ffe0e0' 
        elif "Trap" in meta['name']: bg_color = '#fffacd' 
        elif meta['acc'] > 0.95: bg_color = '#e0fffa' 
        
        box_text = (f"C{cid}: {meta['name']}\n"
                    f"Acc: {meta['acc']:.0%}\n"
                    f"--------------------\n"
                    f"[A]: \"{meta['example_a']}\"")
        
        plt.text(
            cx, cy, 
            box_text, 
            fontsize=7, 
            fontweight='bold', 
            ha='center', va='center',
            bbox=dict(facecolor=bg_color, alpha=0.9, edgecolor='#555', boxstyle='round,pad=0.4')
        )

    plt.title("Cognitive State Map: Performance & Uncertainty Deviation", fontsize=16)
    plt.xlabel("t-SNE Dimension 1")
    plt.ylabel("t-SNE Dimension 2")
    plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', title="Cluster Profile", fontsize=10)
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    run_final_cluster_report()
```
